```python
# we call first a fiducial configuration that could serve us when we want to correct measurments taken at certain fiducial values
parameters_fid = {
'H0' :  67.32,
'omch2' : 0.120088016,
'Omega_cdm0' : 0.120088016/(67.32/100.0)**2,
'ombh2' : 0.022387993,
'Omega_b0' : 0.022387993/(67.32/100.0)**2,
'Omega_k0' : 0.0,
'w0' : -1.0,
'wa' : 0.0,
'ns' : 0.9661,
'mnu' : (0.3158-0.314378999)*93.04*pow(0.6732, 2),
'As' : 2.09175e-9,
'gamma_MG' : 0.545    
}

background_fid = CAMBBackground(
            H0=parameters_fid['H0'], Omega_cdm0=parameters_fid['Omega_cdm0'],
            Omega_b0=parameters_fid['Omega_b0'], Omega_k0=parameters_fid['Omega_k0'],
            w0=parameters_fid['w0'], wa=parameters_fid['wa'],  ns=parameters_fid['ns'],
            As=parameters_fid['As'], mnu=parameters_fid['mnu'],
            gamma_MG=parameters_fid['gamma_MG'])

start = time.time()            
Pk_fid = CAMBLinearPerturbations(background_fid, np.linspace(0.0, 2.0, 80))
end = time.time()
print(f"⏱️ Time elapsed: {end - start:.3f} seconds")
```

```python
# Cosmology parameters
cosmo_parameters = dict(
H0 =  67.32, #67.32
# h = 67.0/100.0,
# omch2 = 0.12, #0.120088016
Omega_cdm0 = 0.120088016/(67.32/100.0)**2,
# ombh2 = 0.022, #0.022387993
Omega_b0 = 0.022387993/(67.32/100.0)**2,
Omega_k0 = 0.0,
w0 = -1.0,
wa = 0.0,
ns = 0.9661, #0.9661
mnu = 0.06, #(0.3158-0.314378999)*93.04*pow(0.6732, 2)
As = 2.09175e-9, #2.09175e-9
gamma_MG=0.0,
)

selection_parameters = dict(
        A_l=52.0,
        B_l=0.9,
        C_l=0.5,
        sig_A_l=0.2,
        sig_B_l=-0.05,
        sig_C_l=0.001,
        sig_lambda_norm=0.9,
        sig_lambda_z=0.1,
        sig_lambda_exponent=0.4,
        sig_z_z=0.025,
        sig_z_lambda=5.0e-6,
)

profile_parameters = dict(
        r_interp=np.logspace(-10, 2.5, 200),
        two_halo="None",
        offcentering=False,
        rms_off=0.0,
        f_off=0.0,
        trunc_fact=3.0,
        zs_max=2.0,
        mean_nz=0.4,
        sigma_nz=0.3,
        alpha_nz=0.4,
)

area = 10313
halo_concentration = 0.1
overdensity_type = 'vir'
CG_like_selection = 'CC_CWL_Cxi2'
CG_xi2_cov_selection = 'covCC_covCxi2'
bias = 'castro23'
neutrino_cdm = False
nonu = False

zed_obs_edges = np.linspace(0.2, 1.8, 9)
Lambda_obs_edges = np.array([20.0, 30.0, 45.0, 60.0, 500.0])
Rad_obs_edges = np.linspace(5.0, 100.0, 11)
Lambda_obs_Cxi2_edges = np.array([20, 30, 500])
Rad_obs_Cxi2_edges = np.geomspace(20.0, 130.0, 31)
zed_obs_Cxi2_edges = np.arange(0.2, 1.81, 0.4)
        
k = np.geomspace(1e-4, 10, 500)
Mass = np.logspace(12.0, 16.0, 51)
Lambda = np.geomspace(5.0, 250.0, 51)
zed   = np.linspace(1.e-5, 6.0-1.e-5, 200)

# Instantiate classes

background = CAMBBackground(**cosmo_parameters)

perturbations = CAMBLinearPerturbations(background, np.linspace(0.0, 2.0, 100))
perturbations_fid = CAMBLinearPerturbations(background, np.linspace(0.0, 2.0, 100))

HS = HaloStatistics(perturbations, z=zed, k=k, overdensity_type=overdensity_type, nonu=nonu)

HSCastro = CastroHMFBias(HS)
profileNFWCastro = ProfileNFW(HSCastro, k=k, z=zed, **profile_parameters)
profileBMOCastro = ProfileBMO(HSCastro, k=k, z=zed, **profile_parameters)

HSTinker = TinkerHMFBias(HS)
profileNFWTinker = ProfileNFW(HSTinker, k=k, z=zed, **profile_parameters)
profileBMOTinker = ProfileBMO(HSTinker, k=k, z=zed, **profile_parameters)

selectionFunction = SelectionFunction(**selection_parameters)

haloClustering = HaloClustering(
        perturbations, perturbations_fid, selectionFunction, k=k
    )

covariance = HaloCovariance(
        perturbations, area=area, nbins_zob=len(zed_obs_edges), k=k
    )

clusterStatistics = ClusterStatistics(
        perturbations,
        HS,
        selectionFunction,
        HSCastro,
        profileNFWCastro,
        haloClustering,
        covariance,
        z_obs_edges=zed_obs_edges,
        Lambda_obs_edges=Lambda_obs_edges,
        Rad_obs_edges=Rad_obs_edges,
        Lambda_obs_Cxi2_edges=Lambda_obs_Cxi2_edges,
        Rad_obs_Cxi2_edges=Rad_obs_Cxi2_edges,
        z_obs_Cxi2_edges=zed_obs_Cxi2_edges,
        halo_concentration=halo_concentration,
        k=k,
        Mass=Mass,
        Lambda=Lambda,
        z=zed,
        area=area,
        CG_like_selection=CG_like_selection,
        CG_xi2_cov_selection=CG_xi2_cov_selection,
        bias=bias,
        neutrino_cdm=neutrino_cdm,
    )

start = time.time()
N_zbin_Lbin, g_zbin_Lbin_Rbin, Cxi2_zbin_Lbin_Rbin, cov_zbin_Lbin, cov_Cxi2 = (
        clusterStatistics.N_zbin_Lbin_Rbin()
    )
end = time.time()
print(f"⏱️ Time elapsed: {end - start:.3f} seconds")
```
